```python
import matplotlib.pyplot as plt

# --- 1. THE MINI-DATASET ---
# List of (Gormati, Hindi) pairs
data = [
    ("Tu kat ja ro chhi", "Tu kaha ja raha hai"),
    ("Maar naam Rajesh chha", "Mera naam Rajesh hai"),
    ("Aaj vatavaran kharaab chha", "Aaj mausam kharaab hai"),
    ("Kayi kar roo cho", "Kya kar rahe ho"),
    ("Tamar baper naam kayi chh", "Tumhare pita ka naam kya hai")
]

# --- 2. THE ANALYSIS LOOP ---
gormati_unique_total = 0
hindi_unique_total = 0

print("Scanning Dataset for Linguistic Gaps...\n")

for g_sent, h_sent in data:
    # Tokenize (Split into words)
    g_vocab = set(g_sent.lower().split())
    h_vocab = set(h_sent.lower().split())
    
    # Calculate Gaps
    # Words in Gormati that are NOT in Hindi
    g_gap = len(g_vocab - h_vocab)
    
    # Words in Hindi that are NOT in Gormati (usually 0 or low overlap)
    h_gap = len(h_vocab - g_vocab) 
    
    # Add to totals
    gormati_unique_total += g_gap
    hindi_unique_total += h_gap
    
    print(f"Sentence Gap: {g_gap} unique words found.")

# --- 3. THE RESEARCH PLOT (AGGREGATE) ---
categories = ['Hindi Context', 'Gormati Unique Context']
counts = [hindi_unique_total, gormati_unique_total]

plt.figure(figsize=(10, 6))
# We use 'Maroon' for Gormati to look academic
plt.bar(categories, counts, color=['lightgray', 'maroon'])

plt.title(f"Aggregate Vocabulary Gap Analysis (N={len(data)} sentences)")
plt.ylabel("Total Count of Untranslatable Tokens")
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Add text on top of the bars
for i, v in enumerate(counts):
    plt.text(i, v + 0.5, str(v), ha='center', fontweight='bold')

# 👉 ADD THIS LINE TO SAVE THE IMAGE 
plt.savefig("vocab_gap.png", dpi=300, bbox_inches='tight')

# plt.show() MUST come AFTER plt.savefig()
plt.show()
```
