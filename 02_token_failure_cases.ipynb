```python
#"Failure Case #1: Llama-3 Tokenization Failure"

import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

# 1. THE DATA (What you just found)
# Let's pretend the AI split it like this (Replace with your actual tokens if different)
# If your output was different, paste it inside the [ ] below.
tokens = ['Tu', 'kat', 'ja', 'ro', 'ch', '##hi'] 

# 2. THE SCORE (1 = Good, 0 = Bad/Broken)
# Tu (1), kat (1), ja (1), ro (1), ch (0.5), ##hi (0.5 - Broken!)
scores = [1, 1, 1, 1, 0.5, 0.5] 

# 3. THE VISUALIZATION
plt.figure(figsize=(10, 2))
sns.heatmap([scores], annot=[tokens], fmt="", cmap="RdYlGn", cbar=False,
            xticklabels=False, yticklabels=False, square=True, linewidths=2)

plt.title("Llama-3 Tokenization Failure on Gormati (Red = Broken)")
plt.savefig("fail_case_1_llama.png", dpi=300, bbox_inches='tight')
plt.show()
```

```python
from transformers import AutoTokenizer
import seaborn as sns
import matplotlib.pyplot as plt

# 1. LOAD THE BRAIN
tokenizer = AutoTokenizer.from_pretrained("google/muril-base-cased")

# 2. THE HARDER TEST CASE
# "Baper" is the killer word here.
gormati_text = "Tamar baper naam kayi chh"

# 3. RUN THE TOKENIZER
tokens = tokenizer.tokenize(gormati_text)

# 4. AUTO-DETECT BROKEN PIECES
# If a token starts with '##', it means the word was shattered.
# We give it a score of 0 (Red). Normal words get 1 (Green).
scores = [0 if "##" in t else 1 for t in tokens]

# 5. PRINT THE DEBUG INFO
print(f"Original: {gormati_text}")
print(f"Tokens:   {tokens}")
print(f"Broken Pieces Detected: {tokens.count('##') + sum(x == 0 for x in scores)}")

# 6. GENERATE THE HEATMAP
plt.figure(figsize=(12, 2))
sns.heatmap([scores], annot=[tokens], fmt="", cmap="RdYlGn", cbar=False,
            xticklabels=False, yticklabels=False, square=True, linewidths=2)

plt.title("Failure Case #2: The 'Suffix' Problem (Red = Shattered)")
plt.savefig("fail_case_2_suffix.png", dpi=300, bbox_inches='tight')
plt.show()
```
