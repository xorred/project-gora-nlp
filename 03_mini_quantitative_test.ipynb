```python
from transformers import AutoTokenizer
import pandas as pd

# 1. SETUP
tokenizer = AutoTokenizer.from_pretrained("google/muril-base-cased")

# 2. YOUR DATASET (The sentences you've built so far)
data = [
    # Your new sentence
    {"gormati": "Hamar Khetem aajo tu ton dakalu chhu", "hindi": "Hamare khet mein aao tumko dikhata hoon"},
    # Previous sentences
    {"gormati": "Tu kat ja ro chhi", "hindi": "Tu kaha ja raha hai"},
    {"gormati": "Tamar baper naam kayi chh", "hindi": "Tumhare pita ka naam kya hai"},
    {"gormati": "Taar tandri acchi chh", "hindi": "Tumhari patni acchi hai"},
    {"gormati": "Kayi kar ro chi", "hindi": "Kya kar rahe ho"}
]

# 3. THE CALCULATOR
results = []
print("Calculating Research Metrics...\n")

for row in data:
    # Gormati Scores
    g_words = len(row['gormati'].split())
    g_tokens = len(tokenizer.tokenize(row['gormati']))
    g_fertility = g_tokens / g_words
    
    # Hindi Scores
    h_words = len(row['hindi'].split())
    h_tokens = len(tokenizer.tokenize(row['hindi']))
    h_fertility = h_tokens / h_words
    
    results.append({
        "Sentence": row['gormati'],
        "Gormati_Score": round(g_fertility, 2),
        "Hindi_Score": round(h_fertility, 2),
        "Inefficiency": round(((g_fertility / h_fertility) - 1) * 100, 1)
    })

# 4. SHOW THE TABLE
df = pd.DataFrame(results)
print(df[['Sentence', 'Gormati_Score', 'Hindi_Score', 'Inefficiency']])

# 5. THE FINAL PAPER STATISTIC
avg_gap = df['Inefficiency'].mean()
print("\n" + "="*40)
print(f"FINAL RESULT FOR ABSTRACT:")
print(f"The model is {round(avg_gap, 1)}% less efficient on Gormati than Hindi.")
print("="*40)
```
