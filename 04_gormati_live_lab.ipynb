```python
from transformers import AutoTokenizer

# 1. SETUP
tokenizer = AutoTokenizer.from_pretrained("google/muril-base-cased")

print("--- GORMATI LIVE LAB ---")
print("Type 'exit' to stop.\n")

while True:
    # 1. Get your input
    g_sent = input("\nEnter Gormati Sentence: ")
    if g_sent.lower() == 'exit':
        break
        
    h_sent = input("Enter Hindi Translation:  ")
    
    # 2. Calculate the Math
    g_tokens = tokenizer.tokenize(g_sent)
    h_tokens = tokenizer.tokenize(h_sent)
    
    g_len = len(g_tokens)
    h_len = len(h_tokens)
    
    # 3. The Score
    # If Gormati uses more tokens, the score goes UP (Good for us)
    # If Score > 1.3, it's a great sentence.
    ratio = g_len / h_len if h_len > 0 else 0
    
    print(f"\n📊 ANALYSIS:")
    print(f"Gormati Tokens ({g_len}): {g_tokens}")
    print(f"Hindi Tokens   ({h_len}): {h_tokens}")
    print(f"⚠️ Inefficiency Score: {round(ratio, 2)}x")
    
    if ratio > 1.3:
        print("✅ GREAT SENTENCE! (High Tokenization Failure)")
    elif ratio < 1.0:
        print("❌ TOO EASY (AI thinks it is Hindi)")
    else:
        print("⚠️ AVERAGE (Try harder words)")
    
    print("-" * 30)
```
