```python
from transformers import AutoTokenizer

# 1. SETUP
tokenizer = AutoTokenizer.from_pretrained("google/muril-base-cased")

print("--- GORMATI TOKENIZATION LAB ---")
print("Type 'exit' to stop.\n")

while True:
    # 1. Get input
    g_sent = input("\nEnter Gormati Sentence: ")
    if g_sent.lower() == 'exit':
        break

    h_sent = input("Enter Hindi Translation:  ")

    # 2. Tokenize
    g_tokens = tokenizer.tokenize(g_sent)
    h_tokens = tokenizer.tokenize(h_sent)

    g_len = len(g_tokens)
    h_len = len(h_tokens)

    # 3. Sentence-level ratio (naive metric — conflates fragmentation with
    #    cross-linguistic verbosity differences; kept for continuity with
    #    earlier pilot results, not the primary metric going forward)
    ratio = g_len / h_len if h_len > 0 else 0

    # 4. Word-normalized fragmentation ratio (subwords per word, per language)
    #    Isolates tokenizer fragmentation from sentence-length effects.
    g_words = len(g_sent.split())
    h_words = len(h_sent.split())
    g_frag = g_len / g_words if g_words > 0 else 0
    h_frag = h_len / h_words if h_words > 0 else 0
    frag_delta = g_frag - h_frag

    print(f"\nANALYSIS:")
    print(f"Gormati Tokens ({g_len}): {g_tokens}")
    print(f"Hindi Tokens   ({h_len}): {h_tokens}")
    print(f"Sentence-level ratio: {round(ratio, 2)}x")
    print(f"Fragmentation delta (subwords/word, Gormati - Hindi): {round(frag_delta, 2)}")

    if frag_delta > 0.3:
        print("Notable fragmentation gap — Gormati words split into substantially more subwords than Hindi.")
    elif frag_delta < 0:
        print("No fragmentation penalty detected for this pair.")
    else:
        print("Modest fragmentation gap.")

    print("-" * 30)
```
